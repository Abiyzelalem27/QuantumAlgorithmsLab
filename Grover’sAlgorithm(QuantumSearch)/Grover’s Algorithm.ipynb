# Grover’s Algorithm: Quantum Search

**Problem statement:** Given $N$ possible candidates, find one that satisfies a specified condition. For example, find a particular name in an **unsorted list**, or find an input $x$ for which $f(x)=1$.

The best search method depends on whether the data is **unstructured, sorted, or indexed**.

## 1. Classical Search in an Unsorted List

Without an index or useful structure, a classical algorithm checks candidates one at a time. This is called **linear search**.

For a list containing exactly one matching element:

- **Best case:** $1$ check—the match is first.
- **Average case:** $(N+1)/2$ checks, assuming the match is equally likely to occupy any position.
- **Worst case:** $N$ checks—the match is last.

If no matching element exists, all $N$ candidates must be checked to establish its absence.

Therefore, classical unstructured search requires

$$
\boxed{\Theta(N)}
$$

checks on average and in the worst case.

## 2. How Grover’s Algorithm Improves Unstructured Search

Grover’s algorithm uses a **quantum oracle** to mark matching candidates by changing their phase. Repeated **amplitude amplification** increases the probability of measuring a matching candidate.

The algorithm starts from the uniform superposition

$$
|s\rangle=\frac{1}{\sqrt{N}}\sum_{x=0}^{N-1}|x\rangle.
$$

The phase oracle acts as

$$
O_f|x\rangle=(-1)^{f(x)}|x\rangle,
$$

where $f(x)=1$ for a matching candidate and $f(x)=0$ otherwise.

Thus:

- Matching candidates acquire a minus sign.
- Nonmatching candidates retain their original phase.

Each Grover iteration applies the oracle followed by the **diffusion operator**

$$
D=2|s\rangle\langle s|-I,
$$

where $I$ is the identity operator.

The combined Grover operator is

$$
G=DO_f.
$$

For exactly one matching candidate, after $r$ iterations the success probability is

$$
P_r=\sin^2\big((2r+1)\theta\big),
\qquad
\sin\theta=\frac{1}{\sqrt{N}}.
$$

Choose a nonnegative integer $r$ close to

$$
r=\frac{\pi}{4\theta}-\frac{1}{2}
$$

to make the success probability close to its first maximum.

For large $N$, this gives approximately

$$
r\approx\frac{\pi}{4}\sqrt{N}.
$$

Therefore, Grover’s algorithm requires

$$
\boxed{\Theta(\sqrt{N})}
$$

oracle queries to find the match with high probability.

This is a **quadratic speedup** over classical unstructured search, not an exponential speedup.

### Example: One Million Candidates

For $N=1{,}000{,}000$ candidates and exactly one match:

| Method | Approximate search effort |
|---|---:|
| Classical linear search, average case | 500,000 checks |
| Classical linear search, worst case | 1,000,000 checks |
| Grover search | 785 Grover iterations |

Each Grover iteration contains one phase-oracle query plus diffusion operations.

**These query counts do not directly determine execution time.** A quantum oracle query may cost much more than a classical check.

## 3. Why Classical Search Can Be Better for Sorted Data

If the list is already sorted by the search key, a classical algorithm can use **binary search**.

Each comparison eliminates approximately half the remaining candidates:

$$
N\rightarrow\frac{N}{2}\rightarrow\frac{N}{4}\rightarrow\cdots.
$$

Binary search requires

$$
\boxed{O(\log_2 N)}
$$

comparisons, assuming efficient access to the middle element.

For one million sorted entries, binary search needs at most about **20 comparisons**. Ordinary Grover search requires approximately **785 iterations** for one match.

Therefore, **classical binary search uses fewer queries than ordinary Grover search for this sorted-search task**.

This comparison concerns ordinary Grover search. It does not imply that every quantum algorithm ignores sorted structure.

**Sorting helps only when the ordering is relevant to the search.** A list sorted by surname does not automatically make searching by an unrelated property efficient.

## 4. Comparing Different Search Situations

| Data or task | Classical approach | Comparison with ordinary Grover search |
|---|---|---|
| Unsorted data with no useful structure | Linear search: $\Theta(N)$ | Grover uses $\Theta(\sqrt{N})$ queries for one match |
| Data sorted by the search key | Binary search: $O(\log N)$ | Binary search uses fewer queries asymptotically |
| Data with a suitable hash index | Expected $O(1)$ lookup under standard assumptions | Grover offers no asymptotic lookup advantage |
| Small search space | Few checks with little overhead | Quantum overhead can outweigh the query savings |
| Candidates generated by a computable rule | Exhaustive search may require $\Theta(N)$ checks if no useful structure is available | Grover can reduce queries if the checking rule has an efficient quantum implementation |

An efficient classical algorithm that exploits additional structure may outperform Grover even when the candidates are not stored in a sorted list.

## 5. Why Not Sort Every Unsorted List First?

Sorting itself takes work. Efficient general-purpose comparison sorting requires

$$
\Theta(N\log N)
$$

comparisons in the worst case.

For **one search**, sorting the entire list first is usually more expensive than performing a linear scan.

For **many searches**, sorting once and then using binary search can be worthwhile.

For $Q$ searches, the worst-case costs are

$$
\text{Repeated linear searches: }O(QN),
$$

and

$$
\text{Sort once, then binary search: }
O(N\log N+Q\log N).
$$

Building a suitable hash index also requires initial work and additional storage, but can provide very fast subsequent lookups.

Therefore, the comparison should include both **preprocessing costs** and **the number of searches performed**.

## 6. Practical Limits of Grover’s Advantage

Grover’s theoretical speedup concerns the **number of oracle queries**.

A practical implementation must also account for:

- Preparing the initial quantum state.
- Implementing the checking function as a coherent quantum circuit.
- Providing coherent access to any required database records.
- Performing diffusion operations.
- Controlling errors during the computation.
- Measuring and verifying the result.

Loading an arbitrary classical database into a quantum-accessible form can require substantial work. This preparation cost may outweigh the savings for a single search.

Grover’s algorithm does **not** reveal all database entries simultaneously. Measurement produces a candidate, and amplitude amplification increases the probability that this candidate is a match.

Also, performing too many Grover iterations can **reduce** the success probability because the amplitudes continue rotating beyond the optimal point.

## 7. Searching When There Are Multiple Matches

If there are $M$ matching candidates, with $1\leq M\leq N$, define

$$
\sin\theta=\sqrt{\frac{M}{N}}.
$$

The probability of measuring any matching candidate after $r$ iterations is

$$
P_r=\sin^2\big((2r+1)\theta\big).
$$

For $M\ll N$, the approximate iteration count is

$$
r\approx\frac{\pi}{4}\sqrt{\frac{N}{M}}.
$$

The quantum query complexity for finding one matching candidate is

$$
\boxed{\Theta\left(\sqrt{\frac{N}{M}}\right)}.
$$

For comparison, classical random sampling requires an expected

$$
\Theta\left(\frac{N}{M}\right)
$$

checks to find a match.

These formulas assume at least one match exists. If the number of matches is unknown, an adapted search procedure is needed; the single-match iteration count should not be applied blindly.

## Key Conclusion

**Grover’s algorithm provides a quadratic query advantage for unstructured search.**

**Classical search can be more efficient when useful ordering, indexing, or other exploitable structure is available.**

The practical choice depends on the search structure, preprocessing requirements, oracle cost, hardware overhead, and number of searches—not just the size of the search space.

### Formal Description

$x$ indexes the entries in the database (a string of $n = \log N$ bits).

$f$ is the “oracle” function that recognizes the desired element $x^*$ (the solution):

$$
f(x) =
\begin{cases}
1, & \text{for } x = x^*, \\
0, & \text{otherwise}.
\end{cases}
$$

For example, $f$ compares the entry with index $x$ to the name we are looking for.

## Examples of Search Problems

### 1. Factoring an Integer

**Problem:** Find a nontrivial divisor $p$ of an integer $m$, knowing that

$$
m=pq,
$$

where $p$ and $q$ are prime numbers.

At least one of these factors must satisfy

$$
p\leq\sqrt{m}.
$$

A brute-force approach checks candidate divisors from $2$ to $\lfloor\sqrt{m}\rfloor$.

Define the checking function

$$
f(p)=
\begin{cases}
1, & \text{if } p \text{ divides } m,\\
0, & \text{otherwise}.
\end{cases}
$$

Classical trial division uses $O(\sqrt{m})$ candidate checks in the worst case.

Applying Grover’s algorithm to this candidate space would require approximately

$$
O\left(\sqrt{\sqrt{m}}\right)
=
O(m^{1/4})
$$

oracle queries, assuming an appropriate quantum divisibility oracle.

**Important:** These are candidate-check or query counts, not total computational costs.

Factoring has mathematical structure that more specialized algorithms can exploit. In particular, **Shor’s algorithm** solves integer factoring in polynomial time in the input bit length on an ideal quantum computer.

If $m$ has $n$ bits, then $m$ is approximately $2^n$. Consequently, the Grover-based bound becomes

$$
O(m^{1/4})=O(2^{n/4}),
$$

which is still exponential in $n$.

### 2. Searching for Witnesses in NP Problems

**NP** is the class of decision problems for which a “yes” answer has a certificate, or **witness**, that can be verified in polynomial time.

Given an instance $x$ and a candidate witness $w$, a verifier checks whether $w$ establishes a “yes” answer:

$$
V(x,w)=
\begin{cases}
1, & \text{if } w \text{ is a valid witness for } x,\\
0, & \text{otherwise}.
\end{cases}
$$

One possible strategy is to search through all candidate witnesses.

For a space containing $N$ candidates and exactly one valid witness:

- Classical unstructured search requires $\Theta(N)$ checks.
- Grover’s algorithm requires $\Theta(\sqrt{N})$ verifier-oracle queries.

For $k$-bit witnesses, there are

$$
N=2^k
$$

possible candidates. Therefore, the search costs are

$$
\text{Classical: }\Theta(2^k),
$$

and

$$
\text{Grover: }\Theta(2^{k/2}).
$$

**Grover provides a quadratic speedup, but the search remains exponential in the witness length.** This does not establish a polynomial-time quantum algorithm for every NP problem.

### General Lesson

Searching all candidates is useful when no exploitable structure is available.

When the problem has useful mathematical or logical structure, specialized algorithms may perform much better than either classical exhaustive search or ordinary Grover search.

## Quantum Oracle

A Boolean function $f:\{0,1\}^n\rightarrow\{0,1\}$ identifies solutions:

$$
f(x)=
\begin{cases}
1, & \text{if } x \text{ is a solution},\\
0, & \text{otherwise}.
\end{cases}
$$

The quantum oracle implements

$$
O_f|x\rangle|y\rangle
=
|x\rangle|y\oplus f(x)\rangle,
$$

where $\oplus$ denotes addition modulo $2$.

### Phase Kickback

Prepare the auxiliary qubit in the state

$$
|-\rangle=\frac{|0\rangle-|1\rangle}{\sqrt{2}}.
$$

Applying the oracle gives

$$
O_f\left(|x\rangle|-\rangle\right)
=
(-1)^{f(x)}|x\rangle|-\rangle.
$$

The auxiliary qubit remains unchanged, while the input acquires a phase:

- If $f(x)=0$, the phase is unchanged.
- If $f(x)=1$, the amplitude acquires a minus sign.

Thus, the oracle **marks solutions through their phase**.

For a superposition of inputs,

$$
O_f\left(\sum_x a_x|x\rangle|-\rangle\right)
=
\sum_x a_x(-1)^{f(x)}|x\rangle|-\rangle.
$$

The oracle alone does not change measurement probabilities. The following diffusion operation converts these relative phase changes into amplitude changes.

### Number of Solutions

Initially, consider exactly one solution $x^*$:

$$
f(x^*)=1,
\qquad
f(x)=0\quad\text{for }x\neq x^*.
$$

Grover’s algorithm also works with $M$ solutions among $N$ candidates. For $1\leq M\ll N$, the approximate number of iterations is

$$
r\approx\frac{\pi}{4}\sqrt{\frac{N}{M}}.
$$

**Knowing $M$ helps choose the iteration count.** However, $M$ does not have to be known beforehand: adapted quantum search procedures can handle an unknown number of solutions.
